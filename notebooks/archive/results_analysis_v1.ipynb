# DPO — Analyse des résultats ETHICS

Charge les JSON produits par `scripts/eval_baseline.py` et `scripts/eval_dpo.py`, calcule les tableaux comparatifs, trace les figures pour le rapport.

Lance ce notebook depuis la racine du projet (`prjgourru/`).

```python
import json
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

RESULTS = Path('../results')
RESULTS.mkdir(exist_ok=True)
list(RESULTS.glob('*.json'))
```

## 1. Helpers de chargement

```python
def load_eval(path: Path) -> dict:
    if not path.exists():
        return None
    with open(path, encoding='utf-8') as f:
        return json.load(f)


def eval_to_df(eval_json: dict, label: str) -> pd.DataFrame:
    if eval_json is None:
        return pd.DataFrame()
    rows = []
    for cat, info in eval_json['details'].items():
        rows.append({
            'model': label,
            'category': cat,
            'accuracy': info['accuracy'],
            'n': info['n'],
        })
    df = pd.DataFrame(rows)
    df['accuracy_pct'] = df['accuracy'] * 100
    return df


def per_class_metrics(eval_json: dict, label: str) -> pd.DataFrame:
    """Per-category accuracy + recall on each gold class + yes_share.

    yes_share = fraction of predictions equal to 1 ("Yes"). Close to 0 means
    the model is biased toward "No"; close to the gold base rate (0.5 when
    balanced) is well-calibrated.
    """
    if eval_json is None:
        return pd.DataFrame()
    rows = []
    for cat, info in eval_json['details'].items():
        preds = np.array(info['predictions'])
        golds = np.array(info['golds'])
        mask1 = golds == 1
        mask0 = golds == 0
        rows.append({
            'model': label,
            'category': cat,
            'accuracy': float((preds == golds).mean()),
            'recall_gold1': float(preds[mask1].mean()) if mask1.any() else float('nan'),
            'recall_gold0': float((1 - preds[mask0]).mean()) if mask0.any() else float('nan'),
            'yes_share': float(preds.mean()),
            'n': len(preds),
        })
    return pd.DataFrame(rows)


def macro_avg(eval_json: dict) -> float:
    if eval_json is None:
        return float('nan')
    return eval_json['summary']['macro_avg']
```

## 2. Chargement des évaluations

On essaie tour à tour : smoke test CPU, baseline GPU, modèle DPO.

```python
smoke = load_eval(RESULTS / 'baseline_cpu_smoke.json')
baseline = load_eval(RESULTS / 'baseline_eval.json')
dpo = load_eval(RESULTS / 'dpo_eval.json')

for name, ev in [('smoke', smoke), ('baseline', baseline), ('dpo', dpo)]:
    if ev is None:
        print(f'{name}: (not found)')
    else:
        print(f'{name}: macro={macro_avg(ev):.3f}  model={ev.get("model")}')
```

## 3. Tableau récapitulatif

```python
frames = []
if smoke:
    frames.append(eval_to_df(smoke, 'CPU smoke (0.5B)'))
if baseline:
    frames.append(eval_to_df(baseline, 'Baseline (1.5B)'))
if dpo:
    frames.append(eval_to_df(dpo, 'DPO (1.5B)'))

df = pd.concat(frames, ignore_index=True) if frames else pd.DataFrame()
pivot = df.pivot(index='category', columns='model', values='accuracy_pct') if not df.empty else df
pivot
```

### Per-class breakdown : accuracy, recall et `yes_share`

Le tableau ci-dessous éclate l'accuracy en :

- `recall_gold1` : taux de bonnes réponses quand le gold est `1` (action wrong / claim reasonable / trait matches selon la catégorie),
- `recall_gold0` : taux de bonnes réponses quand le gold est `0`,
- `yes_share` : fraction des prédictions = `1`. À set équilibré (50/50), un `yes_share` proche de **0** révèle un biais "No" ; proche de **0.5** signifie un modèle calibré.

```python
# Per-class breakdown — surfaces the "No"-bias more clearly than macro accuracy.
class_frames = []
if smoke:
    class_frames.append(per_class_metrics(smoke, 'CPU smoke (0.5B)'))
if baseline:
    class_frames.append(per_class_metrics(baseline, 'Baseline (1.5B)'))
if dpo:
    class_frames.append(per_class_metrics(dpo, 'DPO (1.5B)'))

class_df = pd.concat(class_frames, ignore_index=True) if class_frames else pd.DataFrame()
if not class_df.empty:
    cols = ['model', 'category', 'accuracy', 'recall_gold1', 'recall_gold0', 'yes_share', 'n']
    display(class_df[cols].round(3))
else:
    print('No evaluation data available yet.')
```

```python
if baseline and dpo:
    deltas = {cat: dpo['summary']['per_category'][cat]['accuracy'] - baseline['summary']['per_category'][cat]['accuracy']
              for cat in baseline['summary']['per_category']}
    (pd.Series(deltas, name='delta_DPO_minus_baseline').to_frame() * 100)
```

## 4. Accuracy et biais "No" en parallèle

Deux panneaux : accuracy (gauche) et `yes_share` (droite). Sur un set équilibré, la ligne verte à 0.5 = taux idéal de "Yes". Tout `yes_share` qui en est très éloigné = biais systématique du modèle vers "No" ou "Yes".

```python
if not class_df.empty:
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    # Left panel — accuracy per category per model
    acc_pivot = class_df.pivot(index='category', columns='model', values='accuracy') * 100
    acc_pivot.plot.bar(ax=axes[0], rot=0)
    axes[0].set_title('ETHICS accuracy per category')
    axes[0].set_ylabel('Accuracy (%)')
    axes[0].set_ylim(0, 100)
    axes[0].axhline(50, color='gray', linestyle='--', linewidth=0.7, label='random')
    axes[0].legend(loc='lower right', fontsize=8)

    # Right panel — yes_share per category per model. With a balanced eval set
    # the gold base rate is 0.5; a yes_share far below 0.5 means "No"-bias.
    ys_pivot = class_df.pivot(index='category', columns='model', values='yes_share')
    ys_pivot.plot.bar(ax=axes[1], rot=0)
    axes[1].axhline(0.5, color='green', linestyle='--', linewidth=0.7, label='gold base rate (balanced)')
    axes[1].set_title('Fraction of "Yes" predictions (yes_share)')
    axes[1].set_ylabel('yes_share')
    axes[1].set_ylim(0, 1)
    axes[1].legend(loc='upper right', fontsize=8)

    plt.tight_layout()
    plt.savefig('../results/fig_accuracy.png', dpi=150)
    plt.show()
else:
    print('No evaluation data available yet — run scripts/eval_baseline.py first.')
```

## 5. Confusion par catégorie

Pour chaque modèle dispo, on regarde la matrice 2×2 par catégorie.

```python
def confusion(eval_json: dict, category: str) -> pd.DataFrame:
    det = eval_json['details'][category]
    preds = np.array(det['predictions'])
    golds = np.array(det['golds'])
    mat = np.zeros((2, 2), dtype=int)
    for g, p in zip(golds, preds):
        mat[g, p] += 1
    return pd.DataFrame(mat, index=['gold=0', 'gold=1'], columns=['pred=0', 'pred=1'])


for ev, label in [(baseline, 'baseline'), (dpo, 'dpo'), (smoke, 'smoke')]:
    if ev is None:
        continue
    print(f'== {label} ==')
    for cat in ev['details']:
        print(f'-- {cat} --')
        print(confusion(ev, cat))
        print()
```

Charge `ablations.json` produit par `scripts/run_ablations.py` et trace l'accuracy macro par β / LR / taille.

```python
abl_path = RESULTS / 'ablations.json'
if abl_path.exists():
    with open(abl_path, encoding='utf-8') as f:
        abl = json.load(f)
    abl_df = pd.DataFrame([
        {
            'tag': r['tag'],
            'beta': r['beta'],
            'learning_rate': r['learning_rate'],
            'n_samples': r['max_train_samples'],
            'macro_avg': r['summary']['macro_avg'],
            **{cat: r['summary']['per_category'][cat]['accuracy'] for cat in r['summary']['per_category']},
        }
        for r in abl['runs']
    ])
    display(abl_df.sort_values('macro_avg', ascending=False))
else:
    print('ablations.json not yet generated — run scripts/run_ablations.py first.')
```

```python
if abl_path.exists():
    fig, ax = plt.subplots(figsize=(7, 4))
    by_beta = abl_df.groupby('beta')['macro_avg'].mean()
    by_beta.plot.bar(ax=ax, color='steelblue')
    ax.set_xlabel('DPO β')
    ax.set_ylabel('macro accuracy')
    ax.set_title('Ablation β — macro ETHICS accuracy')
    ax.axhline(macro_avg(baseline) if baseline else 0.5, color='red', linestyle='--', linewidth=0.7, label='baseline')
    ax.legend()
    plt.tight_layout()
    plt.savefig('../results/fig_ablation_beta.png', dpi=150)
    plt.show()
```
